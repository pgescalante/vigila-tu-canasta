```python
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
project_root
```

```python
#Verificar archivos
required_files = [
    "Dockerfile",
    ".dockerignore",
    "requirements.txt",
    "app/app.py",
]

for file in required_files:
    path = project_root / file
    print(f"{file}: {'OK' if path.exists() else 'MISSING'}")
```

```python
# Construir imagen
import subprocess

image_name = "vigila-tu-canasta"

result = subprocess.run(
    ["docker", "build", "-t", image_name, "."],
    cwd=project_root,
    capture_output=True,
    text=True,
)

print(result.stdout)
print(result.stderr)

if result.returncode != 0:
    raise RuntimeError("Docker build failed")
```

```python
# Verificar imagen
result = subprocess.run(
    ["docker", "images", image_name],
    capture_output=True,
    text=True,
)

print(result.stdout)
```

```python
#Correr contenedor
container_name = "vigila-tu-canasta-app"

subprocess.run(
    ["docker", "rm", "-f", container_name],
    capture_output=True,
    text=True,
)

result = subprocess.run(
    [
        "docker", "run",
        "-d",
        "--name", container_name,
        "-p", "8501:8501",
        "-v", f"{Path.home()}/.aws:/root/.aws:ro",
        "-e", "AWS_DEFAULT_REGION=us-east-1",
        image_name,
    ],
    capture_output=True,
    text=True,
)

print(result.stdout)
print(result.stderr)

if result.returncode != 0:
    raise RuntimeError("Docker run failed")
```

```python
# Ver logs
result = subprocess.run(
    ["docker", "logs", container_name],
    capture_output=True,
    text=True,
)

print(result.stdout)
print(result.stderr)
```

```python
# Para deteneder contenedor
subprocess.run(
    ["docker", "rm", "-f", "vigila-tu-canasta-app"],
    capture_output=True,
    text=True,
)
```
